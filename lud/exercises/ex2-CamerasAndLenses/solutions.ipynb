**Ex 2**: Create a function to find distance from lens to where the rays are focused, *b*

```python
def camera_b_distance(f, g):
    """
    camera_b_distance returns the distance (b) where the CCD should be placed
    when the object distance (g) and the focal length (f) are given
    :param f: Focal length
    :param g: Object distance
    :return: b, the distance where the CCD should be placed
    """
    b = 1/((1/f) - (1/g))
    return b
```

```python
f = 15 * math.pow(10,-3)
gs = [0.1, 1, 5, 15]

for g in gs:
    print(f"For g={g}: {camera_b_distance(f,g)}")
```

What is noticable: when $g\rightarrow \infty$, then $b \rightarrow f$
